```python
def assign_points_to_nearest_centroid_1d(data_points, centroids):
    """
    Étape 'Affectation' :
    Pour chaque point, on calcule sa distance à chaque centroïde,
    puis on choisit le centroïde le plus proche.
    
    Retour :
    - point_to_cluster_index : liste de même taille que data_points
      (ex : [0, 0, 1] => le 3e point est dans le cluster 1)
    """
    point_to_cluster_index = []

    for point_value in data_points:
        # Calculer toutes les distances de ce point vers tous les centroïdes
        distances_to_centroids = []
        for centroid_value in centroids:
            distance = abs(point_value - centroid_value)
            distances_to_centroids.append(distance)

        # Trouver l'indice du centroïde le plus proche (distance minimale)
        smallest_distance = min(distances_to_centroids)
        nearest_centroid_index = distances_to_centroids.index(smallest_distance)

        point_to_cluster_index.append(nearest_centroid_index)

    return point_to_cluster_index


def recompute_centroids_as_means_1d(data_points, point_to_cluster_index, number_of_clusters):
    """
    Étape 'Mise à jour' :
    Pour chaque cluster, on récupère les points qui lui appartiennent,
    puis on calcule la moyenne => nouveau centroïde.

    Si un cluster est vide, on met None (cas rare mais possible).
    """
    new_centroids = []

    for cluster_index in range(number_of_clusters):
        # Récupérer les points affectés à ce cluster
        points_in_this_cluster = []
        for i in range(len(data_points)):
            if point_to_cluster_index[i] == cluster_index:
                points_in_this_cluster.append(data_points[i])

        # Calculer la moyenne (si le cluster n'est pas vide)
        if len(points_in_this_cluster) > 0:
            mean_value = sum(points_in_this_cluster) / len(points_in_this_cluster)
            new_centroids.append(mean_value)
        else:
            new_centroids.append(None)

    return new_centroids


def kmeans_1d_simple(data_points, initial_centroids, max_iterations=10):
    """
    K-Means complet :
    Répète Affectation -> Mise à jour jusqu'à convergence (ou max_iterations).
    """
    centroids = initial_centroids[:]  # copie
    number_of_clusters = len(centroids)

    previous_assignment = None

    for iteration in range(max_iterations):

        # Étape 1 : Affectation
        assignment = assign_points_to_nearest_centroid_1d(data_points, centroids)

        # Affichage pédagogique : composition des clusters
        clusters = [[] for _ in range(number_of_clusters)]
        for i in range(len(data_points)):
            cluster_id = assignment[i]
            clusters[cluster_id].append(data_points[i])

        # Convergence : si l'affectation ne change plus, on arrête
        if assignment == previous_assignment:
            break
        previous_assignment = assignment

        # Étape 2 : Mise à jour des centroïdes
        new_centroids = recompute_centroids_as_means_1d(
            data_points=data_points,
            point_to_cluster_index=assignment,
            number_of_clusters=number_of_clusters
        )

        centroids = new_centroids

    return centroids, previous_assignment
```

```python
data_points = [2, 3, 10]       # dataset 1D
initial_centroids = [2, 10]    # K=2 (centres init)

final_centroids, final_assignment = kmeans_1d_simple(
    data_points=data_points,
    initial_centroids=initial_centroids,
    max_iterations=10
)

print("\nRésultat final :")
print("Centroides finaux :", final_centroids)
print("Affectation finale :", final_assignment)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def load_image(path: str) -> np.ndarray:
    img = Image.open(path).convert("RGB")
    return np.array(img, dtype=np.uint8)

img = load_image("pokeball.png")

def showImage(img: np.ndarray):
    plt.imshow(img)
    plt.xlabel("axis 1 (x / colonnes)")
    plt.ylabel("axis 0 (y / lignes)")
    plt.show()

def pixelRouge(img: np.ndarray):
    pixel = img[1, 2]   # pixel en position (ligne 1, colonne 2)
    print(pixel)

def returnImage(img: np.ndarray):
    pixel = img[1, 2]
    patch  = np.zeros((2,2,3), dtype=np.uint8)
    patch[:,:] = pixel

    # Affiche la photo entière
    showImage(patch)

    # Ou affiche un petit crop si tu veux une zone :
    # showImage(img[0:50, 0:50])


def remplacerLesCouleurs(img: np.ndarray):
    # Remplace les pixels rouges par du vert
    img[(img[:,:,0] > 200) & (img[:,:,1] < 50) & (img[:,:,2] < 50)] = [0, 255, 0]
    img[(img[:,:,0] > 200) & (img[:,:,1] < 210) & (img[:,:,2] < 210)] = [201, 255, 191]
    showImage(img)

def inverserLesCouleurs(img: np.ndarray):
    # Inverse les couleurs de l'image
    img = 255 - img
    showImage(img)

def afficherMilieu(img: np.ndarray):
    h, w, _ = img.shape
    pixels = img[h//2 - 1 : h//2 + 1, w//2 - 1 : w//2 + 1]
    showImage(pixels)

def afficherRangée(img: np.ndarray):
    img_ligne = img.reshape(1, -1, 3)
    showImage(img_ligne)

showImage(img)
pixelRouge(img)
returnImage(img)
remplacerLesCouleurs(img)
inverserLesCouleurs(img)
afficherMilieu(img)
afficherRangée(img)
```
